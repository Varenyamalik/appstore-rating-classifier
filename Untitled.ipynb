Build the dataset

```python
import pandas as pd, numpy as np

a = pd.read_csv("AppleStore.csv")
d = pd.read_csv("appleStore_description.csv")
df = a.merge(d[["id", "app_desc"]], on="id", how="left")

df = df[df.rating_count_tot > 0].copy()          # drop unrated apps
df["desc_len"]  = df.app_desc.fillna("").str.len()
df["name_len"]  = df.track_name.str.len()
df["ver_major"] = pd.to_numeric(df.ver.str.extract(r"^(\d+)")[0], errors="coerce").fillna(1)
df["size_mb"]   = df.size_bytes / 1e6
df["is_free"]   = (df.price == 0).astype(int)

df["highly_rated"] = (df.user_rating >= 4.0).astype(int)   # target

num = ["size_mb","price","rating_count_tot","rating_count_ver","sup_devices.num",
       "ipadSc_urls.num","lang.num","vpp_lic","desc_len","name_len","ver_major","is_free"]
cat = ["cont_rating", "prime_genre"]
X, y = df[num + cat], df["highly_rated"]
```

Split and save the test file

```python
from sklearn.model_selection import train_test_split
Xtr, Xte, ytr, yte = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

test_out = Xte.copy(); test_out["highly_rated"] = yte
test_out.to_csv("test_data.csv", index=False)
```

Train, evaluate and save the five models

```python
import os
os.makedirs("model", exist_ok=True)
import joblib
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, roc_auc_score, precision_score,
                             recall_score, f1_score, matthews_corrcoef)

pre = ColumnTransformer([
    ("n", StandardScaler(), num),
    ("c", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat)])

models = {
    "logistic_regression": LogisticRegression(max_iter=1000),
    "decision_tree": DecisionTreeClassifier(max_depth=6, random_state=42),
    "knn": KNeighborsClassifier(),
    "naive_bayes": GaussianNB(),
    "svm": SVC(probability=True, random_state=42),
}

rows = []
for name, clf in models.items():
    pipe = make_pipeline(pre, clf).fit(Xtr, ytr)
    pred, prob = pipe.predict(Xte), pipe.predict_proba(Xte)[:, 1]
    rows.append([name, accuracy_score(yte, pred), roc_auc_score(yte, prob),
                 precision_score(yte, pred), recall_score(yte, pred),
                 f1_score(yte, pred), matthews_corrcoef(yte, pred)])
    joblib.dump(pipe, f"model/{name}.joblib")

print(pd.DataFrame(rows, columns=["Model","Accuracy","AUC","Precision","Recall","F1","MCC"]).round(3))
```
